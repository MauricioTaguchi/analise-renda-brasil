# ETL — Brazil Income Analysis

This notebook prepares the complementary synthetic dataset for analysis:

- reads the source CSV;
- validates its schema and value ranges;
- creates derived analytical columns; and
- exports the clean dataset.

```python
from pathlib import Path

import pandas as pd

RAW_DATA = Path("../data/brazil_income_raw.csv")
CLEAN_DATA = Path("../data/brazil_income_clean.csv")

income = pd.read_csv(RAW_DATA)
income.columns = [column.strip().lower() for column in income.columns]

required_columns = {
    "year",
    "state_code",
    "region",
    "average_income_brl",
    "population",
    "minimum_wage_brl",
}
assert required_columns.issubset(income.columns)
assert income["year"].between(2010, 2100).all()
assert income["average_income_brl"].gt(0).all()
assert income["population"].gt(0).all()

income["income_in_minimum_wages"] = (
    income["average_income_brl"] / income["minimum_wage_brl"]
).round(2)
income["estimated_total_income_brl"] = (
    income["average_income_brl"] * income["population"]
).round(2)

CLEAN_DATA.parent.mkdir(parents=True, exist_ok=True)
income.to_csv(CLEAN_DATA, index=False)
income.head()
```
